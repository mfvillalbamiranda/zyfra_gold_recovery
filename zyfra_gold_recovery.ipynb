## Comprobación que el cálculo de la recuperación sea correcto

Cálculo de la recuperación
Tienes que simular el proceso de recuperación del oro del mineral de oro.

Utiliza la siguiente fórmula para simular el proceso de recuperación:



Donde:

C: <br />
Para saber la recuperación del concentrado rougher → la proporción de oro en el concentrado justo después de la flotación o  <br />
Para saber la recuperación del concentrado final → la proporción de oro después de la purificación.  <br />
Sería rougher.output.concentrate_au <br />
F: <br />
Para saber la recuperación del concentrado rougher → la proporción de oro en la alimentación antes de la flotación <br />
Para saber la recuperación del concentrado final → la proporción de oro en el concentrado justo después de la flotación. <br />
Sería rougher.input.feed_au <br />
T: <br />
Para saber la recuperación del concentrado rougher → la proporción de oro en las colas rougher justo después de la flotación. <br />
Para saber la recuperación del concentrado final → la proporción de oro después de la purificación. <br />
Para predecir el coeficiente, hay que encontrar la proporción de oro en el concentrado y en las colas. Ten en cuenta que tanto el concentrado final como el concentrado rougher tienen importancia. <br />
Sería rougher.output.tail_au

```python
# función para la recuperación
def calculate_recovery(C, F, T):
    recovery = (C * (F - T) / (F * (C - T))) * 100
    return recovery

c = df_train['rougher.output.concentrate_au']
f = df_train['rougher.input.feed_au']
t = df_train['rougher.output.tail_au']

recovery_calculation = calculate_recovery(c, f, t)
recovery_real_value = df_train['rougher.output.recovery']

# cálculo de EAM
mask = recovery_real_value.notna() & recovery_real_value.notna()
mae = mean_absolute_error(recovery_real_value[mask], recovery_calculation[mask])
print(f"Error absoluto medio: {mae} ")
```

El valor del error absoluto medio (EAM) de 9.303e-15 -tendencia al 0- confirma que la implementación de la fórmula de recuperación coincide perfectamente con los datos existentes en el dataset, dejando una base sólida para el entrenamiento del modelo.

## Cambio de la concentración de metales en función de la etapa de purificación

```python
# Función para calcular la concentración de los metales
def analyze_metal_concentrations(df_train):    
    # Create figure for metal concentration changes
    fig, axes = plt.subplots(3, 1, figsize=(6, 10))
    
    metals = ['au', 'ag', 'pb']
    stages = ['rougher.input.feed', 'rougher.output.concentrate', 'final.output.concentrate']
    stage_labels = ['Mezcla mineral', 'Concentrado Rougher', 'Concentrado Final']
    
    for idx, metal in enumerate(metals):
        means = []
        
        # Obtención de concentraciones de cada etapa
        for stage_prefix in stages:
            col = f'{stage_prefix}_{metal}'            
            if col in df_train.columns:
                mean = df_train[col].mean()
                means.append(mean)

        # Gráficos
        x = np.arange(len(stages))
        bars = axes[idx].bar(x, means, 
                           color='#121314',
                           width=0.5,
                           alpha=0.85)
        
        # Personalización de cada subgráfica
        axes[idx].set_title(f'Cambios en concentración {metal.capitalize()}')
        
        axes[idx].set_xticks(x)
        axes[idx].set_xticklabels(stage_labels)
        
        axes[idx].set_ylabel('Concentración (%)')              
                   
        # Adicionar valores en cada barra
        for i, v in enumerate(means):
            axes[idx].text(i, v/2,  # posición en la mitad
                         f'{v:.1f}',
                         ha='center', # alineación horizontal
                         va='center', # alineación vertical
                         color='white',
                         fontsize=10)
        
    plt.tight_layout() # Organización de los gráficos

analyze_metal_concentrations(df_train)
```

Los tres gráficos permiten identificar los cambios de las concentraciones de oro (Au), plata (Ag) y plomo (Pb) a través de las etapas de purificación. El oro presenta un aumento significativo, iniciando en un 8.1, en el concentrado rougher aumenta al 19.8 y en el concentrado final de 44.1 (aumento del 544% frende al valor inicial). La plata fue más variable, aumentando en el concentrado rougher y disminuyendo en el concentrado final (disminución del 41% frente al valor inicial). El plomo tuvo un aumento importante, iniciando en 3.6, aumentando en el concentrado rougher a 7.6 y en el concentrado final de 10.2 (aumento del 283%).

## Distribución de concentraciones totales de todas las sustancias en las diferentes etapas (materia prima, concentrado rougher y concentrado final)

```python
def analyze_total_concentrations(df):
    # definición de las características a evaluar
    stages = ['rougher.input.feed', 'rougher.output.concentrate', 'final.output.concentrate']
    metals = ['au', 'ag', 'pb']
    
    # creación de las gráficas
    fig, axes = plt.subplots(1, len(stages), figsize=(16, 8))
    fig.suptitle('Distribución de la concentración total por etapas', fontsize=18)
    
    outliers_info = {}
    all_concentrations = []
    
    for i, stage in enumerate(stages):
        total_concentration = pd.Series(0, index=df.index)
        
        # sumatoria de todos los metales
        for metal in metals:
            col = f'{stage}_{metal}'
            if col in df.columns:
                total_concentration += df[col].fillna(0)
        
        # cálculo de los cuartiles
        Q1 = total_concentration.quantile(0.25)
        Q3 = total_concentration.quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Filter out outliers
        clean_concentration = total_concentration[
            (total_concentration >= lower_bound) & 
            (total_concentration <= upper_bound)]
        
        all_concentrations.append(clean_concentration)            
        
        # gráfica de cajas y bigotes
        sns.boxplot(y=clean_concentration, ax=axes[i])
        
        axes[i].set_title(f'{stage}')
        axes[i].set_ylabel('Concentraciones totales')
        
        # adición de texto con los datos obtenidos
        stats_text = f'Media: {clean_concentration.mean():.2f}\n'
        stats_text += f'Mediana: {clean_concentration.median():.2f}\n'
        stats_text += f'Q1-Q3: [{Q1:.2f}-{Q3:.2f}]\n'
        stats_text += f'Min: {clean_concentration.min():.2f}\n'
        stats_text += f'Max: {clean_concentration.max():.2f}\n'
        
        
        axes[i].text(0.27, -0.2, stats_text,
                      transform=axes[i].transAxes,
                      bbox=dict(facecolor='white', alpha=0.8),
                      verticalalignment='center')
    
    plt.tight_layout()
```

# Construcción del modelo

## Función para cálculo de valor final de sMAPE

Utiliza estas fórmulas para las métricas de evaluación:





```python
# funciones de las fórmulas necesarias

# función para cálculo de sMAPE
def smape(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    numerator = np.abs(y_true - y_pred)
    denominator = (np.abs(y_true) + np.abs(y_pred)) / 2
    
    # Evitar división por cero
    mask = denominator != 0
    result = np.zeros_like(numerator)
    result[mask] = numerator[mask] / denominator[mask]
    
    return np.mean(result) * 100

# función para cálculo de sMAPE final
def smape_final(y_true, y_pred):
    # Asegurar que son arrays numpy
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # Calcular sMAPE para cada columna
    rougher_smape = smape(y_true[:, 0], y_pred[:, 0])
    final_smape = smape(y_true[:, 1], y_pred[:, 1])
    
    # Combinar con los pesos
    return 0.25 * rougher_smape + 0.75 * final_smape

smape_scorer = make_scorer(smape_final, greater_is_better=False, needs_proba=False)
```

## Entrenamiento de diferentes modelos

```python
# Preparación de los datos
target_rougher = 'rougher.output.recovery'
target_final = 'final.output.recovery'
features_train = df_train.drop(columns= ['date', target_rougher, target_final])
target_train = df_train[[target_rougher, target_final]]

features_train_split, features_val, target_train_split, target_val = train_test_split(
    features_train, target_train, test_size=0.2, random_state=12345)
```

# función para el entrenamiento de los modelos
def model_training(model, features_train, target_train, features_val, target_val):
    model.fit(features_train, target_train)
    predictions = model.predict(features_val)

    smape_score = smape_final(target_val.values, predictions)
    return smape_score

```python
#función para el entrenamiento de modelos con validación cruzada
def model_training_cv(model, features_train, target_train, cv_folds=5):
    """
    Entrena el modelo usando validación cruzada
    """
    # Usar validación cruzada
    scores = cross_val_score(
        model, 
        features_train, 
        target_train, 
        cv=cv_folds, 
        scoring=smape_scorer,
        n_jobs=-1  # Usar todos los procesadores disponibles
    )
    
    # Como greater_is_better=False, los scores son negativos
    # Los convertimos a positivos
    scores = -scores
    
    print(f"sMAPE scores por fold: {scores}")
    print(f"sMAPE promedio: {scores.mean():.4f}")
    print(f"Desviación estándar: {scores.std():.4f}")
    
    return scores.mean(), scores.std()
```

### Regresión lineal

```python
lr_model = LinearRegression()
lr_smape = model_training_cv(lr_model, features_train_split, target_train_split)
```

### Árbol de decisión

```python
tree_model = DecisionTreeRegressor(max_depth=10, random_state=12345)
tree_smape = model_training_cv(tree_model, features_train_split, target_train_split)
```

### Bosque aleatorio

```python
forest_model = RandomForestRegressor(n_estimators=50, max_depth=10, random_state=12345)
forest_smape = model_training_cv(forest_model, features_train_split, target_train_split)
```

## Conclusión

Información importante previo al análisis: sMAPE significa Error Medio Absoluto Porcentual Simétrico. Es una métrica que mide el error entre las predicciones y los valores reales, expresado en porcentaje. La puntuación ideal de sMAPE es 0%, lo que significaría que tu modelo predice perfectamente todos los valores sin ningún error.

Interpretación de los valores:
0%: Predicción perfecta (ideal, pero prácticamente imposible)  <br />
1-5%: Excelente precisión  <br />
5-10%: Muy buena precisión  <br />
10-20%: Buena precisión  <br />
20-50%: Precisión aceptable  <br />
\>50%: Precisión pobre
